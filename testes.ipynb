```python
from automata.finite_words.dfa import DFA
```

```python
Q = {"q0", "q1", "q2", "q3"}
S = {"a", "b"}
transition_table = {
    ("q0", "a"): "q1",
    ("q0", "b"): "q2", 
    ("q1", "a"): "q0", 
    ("q1", "b"): "q3", 
    ("q2", "a"): "q3",
    ("q2", "b"): "q0",
    ("q3", "a"): "q2",
    ("q3", "b"): "q1"}
q0 = "q0"
F = {"q0"}
```

```python
automato = DFA(Q, S, transition_table, q0, F)
```

```python
automato.eval("")
```
