* DFA

$$A = (Q, \Sigma, \delta, q_0, F)$$

```python
from automata.finite_words.dfa import DFA

Q = {"q0", "q1", "q2", "q3"}
S = {"a", "b"}
transition_table = {
    ("q0", "a"): "q1",
    ("q0", "b"): "q2", 
    ("q1", "a"): "q0", 
    ("q1", "b"): "q3", 
    ("q2", "a"): "q3",
    ("q2", "b"): "q0",
    ("q3", "a"): "q2",
    ("q3", "b"): "q1"}
q0 = "q0"
F = {"q0"}

AFN = DFA(Q, S, transition_table, q0, F)
```

* NFA

$$A = (Q, \Sigma, \delta, q_0, F)$$

```python
from automata.finite_words.nfa import NFA

Q = {"q0", "q1", "q2", "q3"}
S = {"a", "b"}
q0 = "q0"
F = {"q3"}
transition_table = {
    ("q0", "a"): {"q1"},
    ("q0", "b"): set(),
    ("q1", "a"): {"q1"},
    ("q1", "b"): {"q1", "q2"},
    ("q2", "a"): set(),
    ("q2", "b"): {"q3"},
    ("q3", "a"): {"q3"},
    ("q3", "b"): {"q3"}
}
```

```python
AFnD = NFA(Q, S, transition_table, q0, F)
```

```python
AFnD.eval("abababbabababababababbaabbababababbbababababbababababbababababbabababab")
```

* NFA $\to$ DFA

```python
from automata.finite_words.conversions import nfa_to_dfa

Q = {"q1", "q2", "q3", "q4"}
F = {"q4"}
S = {"a", "b"}
q0 = "q1"
delta = {
    ("q1", "a"): {"q1"},
    ("q1", "b"): {"q1", "q2"},
    ("q2", "a"): {"q3"},
    ("q3", "a"): {"q4"},
    ("q3", "b"): {"q4"}
}

AFnD = NFA(Q, S, delta, q0, F)
```

```python
AFD = nfa_to_dfa(AFnD)
```
